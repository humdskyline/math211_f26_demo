### 5. Let's find the proportion of commute times longer than 45 minutes
#### a) Write comments that outline the steps needed to find the proportion of commute times longer than 45 minutes.
#### b) Under each comment, write the Python code that carries out that step.  
##### *Run your code one step at a time, checking each result before moving on.  You will probably want to create a new variable for each step*

```python
# get the people with long commutes
long_commuters = first_survey.where("commute_min", are.above(45))
print(long_commuters)

#find the number of people with long commutes
num_long_commuters = long_commuters.num_rows
print(num_long_commuters)

#find the number of people in the survey
everyone = first_survey.num_rows


long_comm_prop = num_long_commuters / everyone
long_comm_prop
```

### 6. Let's find the 3rd longest commute time

```python
# sort the table by the commute time column so the longest times are at the top (descending)
sorted_by_col_desc = first_survey.sort("commute_min", descending=True)

#get the sorted column
sorted_col_desc = sorted_by_col_desc.column("commute_min")

# get the third one at index 2
third_long_comm = sorted_col_desc.item(2)
third_long_comm
```

### 7. Let's find the 5th shortest commute time

```python
# sort the table by the commute time column so the shortest times are at the top (ascending)
sorted_by_col_asc = first_survey.sort("commute_min", descending=False)

#get the sorted column
sorted_col_asc = sorted_by_col_asc.column("commute_min")

# get the fifth one at index 4
fifth_short_comm = sorted_col_asc.item(4)
fifth_short_comm
```
