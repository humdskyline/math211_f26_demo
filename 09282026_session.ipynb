### Let's roll a single die

```python
die = np.arange(1, 7)
print(die)

#here's the code to pick one value at random from the array die
roll = np.random.choice(die)
print(roll)
```

### What if I wanted to roll the die 3 times?

```python
roll1 = np.random.choice(die)
roll2 = np.random.choice(die)
roll3 = np.random.choice(die)

print("I rolled " + str(roll1) + ", " + str(roll2) + ", and " + str(roll3))
```
